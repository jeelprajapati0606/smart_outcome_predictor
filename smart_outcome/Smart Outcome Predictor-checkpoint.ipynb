### Q6. Identify features and target variables for both tasks.


- Features (X): Student demographic indicators, time spent, sessions, quiz attempts, assessment scores, etc.



- Classification target (y): completion_status — 0 means Not Completed, 1 means Completed.



- Regression target (y): final_score — student ka final score, 0–100.

- 80% Training Data: Used to train the model.
  
- 20% Testing Data: Used to check the model’s performance on new data.

### Q8. Apply basic preprocessing (scaling/encoding if required).

## Dataset Load aur Preparation

```python


import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.impute import SimpleImputer
from sklearn.ensemble import BaggingClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score

df = pd.read_csv("Smart_Outcome_Predictor_Dataset_5200 - Smart_Outcome_Predictor_Dataset_5200.csv.csv")

# Dataset check
print(df.head())

```

```python
print(df.shape)
```

```python
print(df.columns)
```

```python
print(df.isnull().sum())
```

```python
print("Duplicate rows:", df.duplicated().sum())
```

```python
# Target columns check
print(df[["completion_status", "final_score"]].head())

# Features aur targets
X = df.drop(columns=["completion_status", "final_score"])
y_class = df["completion_status"]
y_reg = df["final_score"]
```

```python
# Train-test split: 80% training, 20% testing
X_train, X_test, y_class_train, y_class_test, y_reg_train, y_reg_test = (
    train_test_split(
        X, y_class, y_reg,
        test_size=0.2,
        random_state=42,
        stratify=y_class
    )
)
```

```python
# Categorical columns ko numerical columns mein convert karo
X_train = pd.get_dummies(X_train)
X_test = pd.get_dummies(X_test)

# Train aur test mein same columns rakho
X_train, X_test = X_train.align(
    X_test, join="left", axis=1, fill_value=0
)
```

```python
# Missing values fill karo
imputer = SimpleImputer(strategy="median")

X_train = pd.DataFrame(
    imputer.fit_transform(X_train),
    columns=X_train.columns,
    index=X_train.index
)

X_test = pd.DataFrame(
    imputer.transform(X_test),
    columns=X_test.columns,
    index=X_test.index
)

print("Training data:", X_train.shape)
print("Testing data:", X_test.shape)
```

# Part C: Bagging (Bootstrap Aggregating)

## Q9. Implement Bagging Classifier for course completion prediction.

```python
bag_clf = BaggingClassifier(
    estimator=DecisionTreeClassifier(),
    n_estimators=50,
    random_state=42,
    n_jobs=-1
)

bag_clf.fit(X_train, y_class_train)

pred_class = bag_clf.predict(X_test)

print("Bagging Classifier Accuracy:",
      accuracy_score(y_class_test, pred_class))
```

## Q10. Implement Bagging Regressor for final score prediction.

```python
from sklearn.ensemble import BaggingRegressor
from sklearn.tree import DecisionTreeRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

bag_reg = BaggingRegressor(
    estimator=DecisionTreeRegressor(),
    n_estimators=50,
    random_state=42,
    n_jobs=-1
)

bag_reg.fit(X_train, y_reg_train)

pred_reg = bag_reg.predict(X_test)

print("MAE:", mean_absolute_error(y_reg_test, pred_reg))
print("RMSE:", np.sqrt(mean_squared_error(y_reg_test, pred_reg)))
print("R2 Score:", r2_score(y_reg_test, pred_reg))
```

## Q14. Analyze how weak learners improve sequentially.

```python
import matplotlib.pyplot as plt
from sklearn.ensemble import AdaBoostClassifier
from sklearn.metrics import accuracy_score

estimators = [10, 30, 50, 100]
accuracies = []

for n in estimators:
    model = AdaBoostClassifier(
        n_estimators=n,
        random_state=42
    )

    model.fit(X_train, y_class_train)
    pred = model.predict(X_test)

    accuracies.append(accuracy_score(y_class_test, pred))

plt.plot(estimators, accuracies, marker="o")
plt.xlabel("Number of Estimators")
plt.ylabel("Accuracy")
plt.title("AdaBoost: Number of Estimators vs Accuracy")
plt.grid(True)
plt.show()

for n, acc in zip(estimators, accuracies):
    print(n, "Estimators:", round(acc, 4))
```

## Q17. Analyze learning rate and number of estimators.

```python
from sklearn.ensemble import GradientBoostingClassifier
from sklearn.metrics import accuracy_score
import matplotlib.pyplot as plt

learning_rates = [0.01, 0.05, 0.1, 0.2]
estimators = [50, 100, 150, 200]

results = []

for lr in learning_rates:
    for n in estimators:
        model = GradientBoostingClassifier(
            learning_rate=lr,
            n_estimators=n,
            random_state=42
        )

        model.fit(X_train, y_class_train)
        pred = model.predict(X_test)

        acc = accuracy_score(y_class_test, pred)

        results.append({
            "Learning Rate": lr,
            "Estimators": n,
            "Accuracy": acc
        })

result_df = pd.DataFrame(results)
print(result_df)

for lr in learning_rates:
    temp = result_df[result_df["Learning Rate"] == lr]
    plt.plot(temp["Estimators"], temp["Accuracy"],
             marker="o", label=f"LR={lr}")

plt.xlabel("Number of Estimators")
plt.ylabel("Accuracy")
plt.title("Learning Rate vs Estimators")
plt.legend()
plt.grid(True)
plt.show()
```

# Part E: Voting & Stacking Ensembles

## Q24. Implement Voting Classifier using multiple base classifiers.

```python
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier, VotingClassifier
from sklearn.metrics import accuracy_score
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler

# Models
model1 = make_pipeline(
    StandardScaler(),
    LogisticRegression(max_iter=3000)
)

model2 = DecisionTreeClassifier(random_state=42)

model3 = RandomForestClassifier(
    n_estimators=50,
    random_state=42
)

# Voting Classifier
voting_clf = VotingClassifier(
    estimators=[
        ("lr", model1),
        ("dt", model2),
        ("rf", model3)
    ],
    voting="hard"
)

# Train model
voting_clf.fit(X_train, y_class_train)

# Prediction
voting_pred = voting_clf.predict(X_test)

# Accuracy
print("Voting Classifier Accuracy:",
      accuracy_score(y_class_test, voting_pred))
```

## Q25. Compare Hard Voting vs Soft Voting.

- Hard Voting: Models ki predicted classes mein majority class select hoti hai.
- Soft Voting: Models ki predicted probabilities average karke final class choose hoti hai.

```python
from sklearn.ensemble import VotingClassifier, RandomForestClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler

def create_models(voting_type):
    return VotingClassifier(
        estimators=[
            ("lr", make_pipeline(
                StandardScaler(),
                LogisticRegression(max_iter=3000)
            )),
            ("dt", DecisionTreeClassifier(random_state=42)),
            ("rf", RandomForestClassifier(
                n_estimators=50,
                random_state=42
            ))
        ],
        voting=voting_type
    )

# Hard Voting
hard_model = create_models("hard")
hard_model.fit(X_train, y_class_train)
hard_pred = hard_model.predict(X_test)

# Soft Voting
soft_model = create_models("soft")
soft_model.fit(X_train, y_class_train)
soft_pred = soft_model.predict(X_test)

print("Hard Voting Accuracy:",
      accuracy_score(y_class_test, hard_pred))

print("Soft Voting Accuracy:",
      accuracy_score(y_class_test, soft_pred))
```

## Q26. Implement Stacking Classifier with a meta-learner.

```python
from sklearn.ensemble import StackingClassifier, RandomForestClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score

stack_clf = StackingClassifier(
    estimators=[
        ("dt", DecisionTreeClassifier(random_state=42)),
        ("rf", RandomForestClassifier(
            n_estimators=50, random_state=42
        ))
    ],
    final_estimator=LogisticRegression(max_iter=1000),
    cv=5
)

stack_clf.fit(X_train, y_class_train)

stack_pred = stack_clf.predict(X_test)

print("Stacking Classifier Accuracy:",
      accuracy_score(y_class_test, stack_pred))
```

## Q27. Implement Stacking Regressor for score prediction.

```python
from sklearn.ensemble import StackingRegressor, RandomForestRegressor
from sklearn.tree import DecisionTreeRegressor
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

stack_reg = StackingRegressor(
    estimators=[
        ("dt", DecisionTreeRegressor(random_state=42)),
        ("rf", RandomForestRegressor(
            n_estimators=50, random_state=42
        ))
    ],
    final_estimator=LinearRegression(),
    cv=5
)

stack_reg.fit(X_train, y_reg_train)

stack_reg_pred = stack_reg.predict(X_test)

print("MAE:", mean_absolute_error(y_reg_test, stack_reg_pred))
print("RMSE:", np.sqrt(mean_squared_error(y_reg_test, stack_reg_pred)))
print("R2 Score:", r2_score(y_reg_test, stack_reg_pred))
```

# Part F: Model Evaluation & Comparison

### Q28. Evaluate classification models using Accuracy, Precision, Recall, F1-Score and ROC-AUC.

- Accuracy: Total predictions mein kitni correct hain.
- Precision: Model ne jitne students Completed predict kiye, unmein kitne actually completed the.
- Recall: Actually completed students mein se kitne model ne identify kiye.
- F1-Score: Precision aur Recall ka balanced measure.
- ROC-AUC: Model Completed aur Not Completed students ko kitna achha distinguish karta hai.

```python
from sklearn.metrics import (
    accuracy_score, precision_score, recall_score,
    f1_score, roc_auc_score
)

# Apne trained models
classification_models = {
    "Bagging": bag_clf,
    "AdaBoost": ada_clf,
    "Gradient Boosting": gb_clf,
    "LightGBM": lgbm_clf,
    "XGBoost": xgb_clf,
    "Voting": voting_clf,
    "Stacking": stack_clf
}

classification_results = []

for name, model in classification_models.items():
    pred = model.predict(X_test)

    # ROC-AUC ke liye predicted probability
    if hasattr(model, "predict_proba"):
        probability = model.predict_proba(X_test)[:, 1]
        auc = roc_auc_score(y_class_test, probability)
    else:
        auc = np.nan

    classification_results.append({
        "Model": name,
        "Accuracy": accuracy_score(y_class_test, pred),
        "Precision": precision_score(
            y_class_test, pred, zero_division=0
        ),
        "Recall": recall_score(
            y_class_test, pred, zero_division=0
        ),
        "F1 Score": f1_score(
            y_class_test, pred, zero_division=0
        ),
        "ROC-AUC": auc
    })

classification_results = pd.DataFrame(classification_results)

print(classification_results.round(4))
```

## Q29. Evaluate regression models using MAE, RMSE and R² Score.

- MAE: Actual aur predicted score ka average absolute difference.
- RMSE: Prediction errors ka square-root mean squared measure; bade errors ko zyada weight deta hai.
- R² Score: Model target score ke variation ko kitna explain karta hai. Higher is generally better.

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

regression_models = {
    "Bagging": bag_reg,
    "AdaBoost": ada_reg,
    "Gradient Boosting": gb_reg,
    "LightGBM": lgbm_reg,
    "XGBoost": xgb_reg,
    "Stacking": stack_reg
}

regression_results = []

for name, model in regression_models.items():
    pred = model.predict(X_test)

    regression_results.append({
        "Model": name,
        "MAE": mean_absolute_error(y_reg_test, pred),
        "RMSE": np.sqrt(mean_squared_error(y_reg_test, pred)),
        "R2 Score": r2_score(y_reg_test, pred)
    })

regression_results = pd.DataFrame(regression_results)

print(regression_results.round(4))
```

## Q30. Compare all ensemble techniques and identify the best performing model for each task.

```python
# Classification: highest Accuracy
best_classification = classification_results.sort_values(
    by="Accuracy", ascending=False
).iloc[0]

# Regression: lowest RMSE
best_regression = regression_results.sort_values(
    by="RMSE", ascending=True
).iloc[0]

print("Best Classification Model:",
      best_classification["Model"])

print("Best Classification Accuracy:",
      round(best_classification["Accuracy"], 4))

print()

print("Best Regression Model:",
      best_regression["Model"])

print("Best Regression RMSE:",
      round(best_regression["RMSE"], 4))

print("Best Regression R2:",
      round(best_regression["R2 Score"], 4))
```
